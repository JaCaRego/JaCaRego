Vamos trabalhar nessa aula com uma base de dados que contém os 25 esportistas mais bem pagos segundo a forbes, para 2023. Os dados dos ganhos são em Reais e na casa dos milhões por ano.



*   **Célula `h5bzsaKOxXN4`**: `df = pd.read_csv("ESPORTISTAS_MOD7 (1).csv", delimiter=';')` e `df.head(10)`
    *   **Explicação**: A primeira linha lê o seu arquivo CSV (que é como uma planilha) chamado `ESPORTISTAS_MOD7 (1).csv` e o guarda em uma "tabela" especial do pandas chamada `DataFrame`, que nomeamos como `df`. O `delimiter=';'` diz ao pandas que as colunas no seu arquivo estão separadas por ponto e vírgula. A segunda linha, `df.head(10)`, mostra as 10 primeiras linhas da sua tabela `df`, para você dar uma olhada rápida nos dados.

```python
df = pd.read_csv("ESPORTISTAS_MOD7 (1).csv", delimiter=';')

df.head(10)
```

Antes de começarmos com os conceitos estatísticos é interessante conhecermos nossa base.

*   **Célula `8bE4jYRizBWy`**: `df.groupby('Esporte')['Atleta'].count().reset_index().sort_values(by='Atleta', ascending=False)`
    *   **Explicação**: Esta linha complexa serve para descobrir quantos atletas existem em cada esporte e qual esporte tem mais atletas.
        *   `df.groupby('Esporte')`: Agrupa todos os atletas pelo tipo de 'Esporte' que praticam.
        *   `['Atleta'].count()`: Para cada grupo de esporte, ele conta quantos 'Atletas' existem.
        *   `.reset_index()`: Transforma o resultado de volta em uma tabela.
        *   `.sort_values(by='Atleta', ascending=False)`: Organiza essa nova tabela para mostrar os esportes com mais atletas primeiro (do maior para o menor).

```python
## Vamos primeiro entender quais são os esportes que temos quantos jogadores pertecem a cada um deles
df.groupby('Esporte')['Atleta'].count().reset_index().sort_values(by='Atleta', ascending=False)

# Sempre bom relembramos a utilidade de cada passo do nosso código acima que usaremos durante a aula:
# Usamos o group by para agrupar os dados do df pela coluna 'Esporte'
# O count() contar o número de ocorrências da coluna 'Atleta'
# .reset_index() Resetar o índice do resultado anterior
# .sort_values ordenar os resultados do DataFrame pelo número de atletas
```

Bom, já sabemos que dos 25 atletas 8 deles são de Basquete, sendo esse o esporte que mais aparece, em seguida temos o futebol americano.

*   **Célula `9gVqWRx60rYc`**: `contagem_por_esporte = df['Esporte'].value_counts(normalize=True).reset_index()`
    *   **Explicação**: Este comando calcula a porcentagem de atletas para cada esporte.
        *   `df['Esporte'].value_counts(normalize=True)`: Conta quantas vezes cada esporte aparece e divide pelo total, dando a proporção (ou porcentagem).
        *   `.reset_index()`: Coloca o resultado em um formato de tabela.

```python
# Dessa forma abaixo trazemos a porcentagem de cada esporte
contagem_por_esporte = df['Esporte'].value_counts(normalize=True).reset_index()
contagem_por_esporte
```

# Aplicando os conceitos de estatística:

# 1 - MÉDIA

Antes de começarmos utilizando as funções do pandas para calcularmos, iremos aplicar as formulas de forma manual, iniciando pela formula da média.


*   **Célula `dschx2c91D3B`**: Este bloco calcula a média dos ganhos totais "na mão".
    *   `soma_ganhos_totais = df['Ganhos_Totais'].sum()`: Soma todos os valores da coluna 'Ganhos_Totais'.
    *   `quantidade_de_registros = len(df['Ganhos_Totais'])`: Conta quantos atletas (quantas linhas) existem na coluna 'Ganhos_Totais'.
    *   `media_ganhos_totais = soma_ganhos_totais / quantidade_de_registros`: Divide a soma dos ganhos pelo número de atletas para encontrar a média.

```python
# Criariamos uma variavel chamada soma_ganhos_totais e em seguida aplicamos o SUM para que seja feita a soma da coluna.
soma_ganhos_totais = df['Ganhos_Totais'].sum()
## Em seguida temos que calcular quantos atletas nós temos e para isso usamos o LEN
quantidade_de_registros = len(df['Ganhos_Totais'])
## Para finalizar dividimos a soma da coluna de ganhos pela quantidade de registros de atletas
media_ganhos_totais = soma_ganhos_totais / quantidade_de_registros

print("Média de ganhos totais:", media_ganhos_totais)
```

Uma forma muito mais fácil e funcional é utilizando a função mean:

*   **Célula `HYtJ9p7nkL4i`**: `df['Ganhos_Totais'].mean()`
    *   **Explicação**: Esta é a maneira mais fácil de calcular a média! O `mean()` é uma função já pronta do pandas que faz todo o trabalho de somar e dividir automaticamente para a coluna 'Ganhos_Totais'.

```python
df['Ganhos_Totais'].mean()
```

Agora traremos os ganhos totais por esporte utilizando mean:

*   **Célula `xJinV9pFkygI`**: `df.groupby('Esporte')['Ganhos_Totais'].mean().reset_index().sort_values(by='Ganhos_Totais', ascending=False)`
    *   **Explicação**: Semelhante ao que fizemos para contar atletas, este comando calcula a média dos ganhos para *cada esporte* e depois os organiza do maior para o menor, mostrando quais esportes têm os ganhos médios mais altos.

```python
df.groupby('Esporte')['Ganhos_Totais'].mean().reset_index().sort_values(by='Ganhos_Totais', ascending=False)
#Notem que primeiro realizamos o agrupamento por esporte, aplicando o mean ao meio e ao final fizemos a ordenação.
```

# 2 - MEDIANA

Para calcularmos a mediana também temos uma função especial, a função median que ao ser aplicada funciona exatamente igual a média.  

*   **Célula `RaxUeFjZoeoQ`**: `df['Ganhos_Totais'].median()`
    *   **Explicação**: O comando `median()` calcula a mediana da coluna 'Ganhos_Totais'. A mediana é o valor do meio quando você coloca todos os ganhos em ordem, do menor para o maior. É útil porque não é tão afetada por valores extremos (muito altos ou muito baixos) quanto a média.

```python
df['Ganhos_Totais'].median()
```

Reparem que nossa mediana é menor que nossa média que foi de 377.
Isso nos evidencia que temos uma distribuição dos dados não tão uniforme pois alguns valores tem puxado a média para um valor acima da mediana.

*   **Célula `FdTKo9FIo50j`**: `df.groupby('Esporte')['Ganhos_Totais'].median().reset_index().sort_values(by='Ganhos_Totais', ascending=False)`
    *   **Explicação**: Calcula a mediana dos ganhos para cada esporte e organiza os resultados.

```python
df.groupby('Esporte')['Ganhos_Totais'].median().reset_index().sort_values(by='Ganhos_Totais', ascending=False)
```

# 3 - MODA

Nesse caso podemos utilizar a moda para sabermos se temos um total de ganhos que se repete mas não é uma métrica tão eficiente para essa análise em específico.

*   **Célula `NtaRAv8UqcPg`**: `df['Ganhos_Totais'].mode()`
    *   **Explicação**: O comando `mode()` encontra o(s) valor(es) que mais aparece(m) na coluna 'Ganhos_Totais'. No seu caso, ele retornou vários valores porque alguns valores de ganhos se repetem o mesmo número de vezes.

```python
df['Ganhos_Totais'].mode()
```

# 4 - DESVIO PADRÃO

Quando compararmos a média com a mediana notamos que temos uma distribuição de dados não tao uniforme, através do desvio poderemos identificar qual dos esportes tem maior dispersão entre os ganhos:

Para o desvio padrão por esporte iremos agrupar os dados e em seguida aplicar o std.

*   **Célula `drW_mXxerNv8`**: `desvio_padrao_por_esporte = df.groupby('Esporte')['Ganhos_Totais'].std().reset_index()`
    *   **Explicação**: O comando `std()` calcula o desvio padrão dos ganhos para cada esporte. O desvio padrão nos diz o quão "espalhados" os dados estão em torno da média. Um desvio padrão alto significa que os ganhos variam muito dentro daquele esporte, enquanto um desvio padrão baixo significa que os ganhos são mais parecidos entre os atletas daquele esporte.

```python
desvio_padrao_por_esporte = df.groupby('Esporte')['Ganhos_Totais'].std().reset_index()
desvio_padrao_por_esporte
```

Notem que alguns registros estão NAN, esses registros ocorrem quando temos apenas 1 registro para o esporte, pois não temos como calcular o desvio padrão se não temos mais de um registro.

Notamos que o desvio para basquete e para Tenis são muito elevados. Enquanto que para Futebol americano temos o menor desvio, que indica que os atletas tendem a ter os ganhos equiparados.

**Dica: Describe**

O método describe() em Python é usado para gerar estatísticas descritivas de resumo de um DataFrame ou de uma série de dados, fornecendo informações úteis sobre a distribuição dos dados.

*   **Célula `_6URFgD_LtcU`**: `df.describe()`
    *   **Explicação**: Esta é uma função super útil! `describe()` gera um resumo estatístico rápido de todas as colunas numéricas da sua tabela `df`. Ele mostra automaticamente:
        *   `count`: Quantos dados não vazios existem.
        *   `mean`: A média.
        *   `std`: O desvio padrão.
        *   `min`: O menor valor.
        *   `25%`, `50%` (mediana), `75%`: Os quartis, que dividem os dados em quatro partes iguais.
        *   `max`: O maior valor.
        É uma ótima maneira de ter uma visão geral rápida dos seus dados!

```python
df.describe()
```
